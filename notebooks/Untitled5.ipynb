```python
flight_df = flight_df.merge(airport_df, left_on="Departure", right_on="Name", how="left")
flight_df = flight_df.rename(columns={"Latitude": "Dep_Lat", "Longitude": "Dep_Lon"})

flight_df = flight_df.merge(airport_df, left_on="Arrival", right_on="Name", how="left")
flight_df = flight_df.rename(columns={"Latitude": "Arr_Lat", "Longitude": "Arr_Lon"})
```

```python
flight_df
```

```python
flight_df = flight_df.drop(columns=['Arrival Time','LocationID_x','Name_x','Address_x','City_x','LocationID_y','Name_y','Address_y','City_y'])
```

```python
from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Radius of the Earth in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    return R * c

flight_df["Distance_km"] = flight_df.apply(lambda row: haversine(row["Dep_Lat"], row["Dep_Lon"], row["Arr_Lat"], row["Arr_Lon"]), axis=1)
```

```python
flight_speeds = {
    "Akasa Air": 820,
    "Air India": 840,
    "IndiGo": 830,
    "SpiceJet": 820,
    "AirAsia India": 810,
    "GoAir": 815
}
```

```python
flight_df["Speed_kmh"] = flight_df["Flight_agency"].map(flight_speeds)
flight_df["EstimatedDuration_hr"] = flight_df["Distance_km"] / flight_df["Speed_kmh"]

def convert_to_hhmm(duration_hr):
    hours = int(duration_hr)
    minutes = int((duration_hr - hours) * 60)
    return f"{hours:02d}:{minutes:02d}"

flight_df["EstimatedDuration_HHMM"] = flight_df["EstimatedDuration_hr"].apply(convert_to_hhmm)
```

```python
flight_df = flight_df.drop(columns={'EstimatedDuration_hr'})
```

```python
def extract_time(df, datetime_col, new_col="Time_HHMM"):
    """
    Extracts time in HHMM format from a datetime column.
    
    Args:
    df (pd.DataFrame): Input DataFrame
    datetime_col (str): Column name containing datetime values
    new_col (str): Name of the new column to store extracted time

    Returns:
    pd.DataFrame: DataFrame with new time column in HHMM format
    """
    df[new_col] = pd.to_datetime(df[datetime_col]).dt.strftime('%H%M')
    return df

x = extract_time(x, "Departure_date", "Departure_time")
```
