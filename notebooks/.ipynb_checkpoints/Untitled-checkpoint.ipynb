```python
import requests
import math

def get_city_coordinates(city_name):
    """Fetches the coordinates of a given city using the GeoNames API."""
    username = 'tisha_agrawal_08'  # Replace with your GeoNames username
    url = f'http://api.geonames.org/searchJSON?q={city_name}&maxRows=1&username={username}'
    
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        if data['totalResultsCount'] > 0:
            city = data['geonames'][0]
            latitude = float(city['lat'])
            longitude = float(city['lng'])
            print(f"Coordinates of {city_name}: {latitude}, {longitude}")
            return latitude, longitude
        else:
            print(f"Could not find coordinates for {city_name}.")
    else:
        print(f"Error fetching city data: {response.status_code}")
    
    return None, None

def haversine(lat1, lon1, lat2, lon2):
    """Calculates the great-circle distance between two points using the Haversine formula."""
    R = 6371  # Radius of Earth in km
    
    # Convert degrees to radians
    lat1, lon1, lat2, lon2 = map(math.radians, [lat1, lon1, lat2, lon2])
    
    # Haversine formula
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = math.sin(dlat / 2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(dlon / 2)**2
    c = 2 * math.asin(math.sqrt(a))
    
    return R * c  # Distance in km

def calculate_flight_duration(distance, speed=250):
    """Calculates estimated flight duration in hours (default speed: 900 km/h)."""
    duration = distance / speed  # Time = Distance / Speed
    hours = int(duration)
    minutes = int((duration - hours) * 60)
    return hours, minutes
```

```python
from datetime import datetime, timedelta

def generate_random_time():
    hour = np.random.randint(0, 24)  # Random hour (0-23)
    minute = np.random.randint(0, 60)  # Random minute (0-59)
    return f"{hour:02}:{minute:02}"

# Convert Flight Duration to timedelta and compute arrival time
def calculate_arrival_time(departure_time, duration):
    try:
        dep_time = datetime.strptime(departure_time, "%H:%M")
        duration_parts = list(map(int, duration.split(":")))
        flight_duration = timedelta(hours=duration_parts[0], minutes=duration_parts[1])
        arrival_time = dep_time + flight_duration
        return arrival_time.strftime("%H:%M")
    except Exception:
        return None  # Handle any potential errors

# Generate departure times and calculate arrival times
India_datas["Departure Time"] = [generate_random_time() for _ in range(len(India_datas))]
India_datas["Arrival Time"] = India_datas.apply(lambda row: calculate_arrival_time(row["Departure Time"], row["Flight Duration (hh:mm)"]), axis=1)
```

```python
import pandas as pd

# Function to convert flight duration from hh:mm to total minutes
def convert_duration_to_minutes(duration):
    hours, minutes = map(int, duration.split(":"))
    return hours * 60 + minutes

# Function to calculate flight price
def calculate_flight_price(row):
    base_price = 500  # Base price for any flight

    # Adjusting price based on flight type
    type_multiplier = {
        "economic": 1.5,
        "premium": 1.8,
        "business": 2.1,
        "firstClass": 2.5
    }
    
    flight_type = row["flightType"]
    multiplier = type_multiplier.get(flight_type, 1.0)
    
    # Distance-based cost (per km charge)
    distance_cost = row["Flight Distance (km)"] * 2
    
    # Duration-based cost (per minute charge)
    duration_cost = convert_duration_to_minutes(row["Flight Duration (hh:mm)"]) * 1.5
    
    # Airline surcharge (mock values for different airlines)
    agency_surcharge = {
        "IndiGo": 100,
        "SpiceJet": 120,
        "GoAir": 110,
        "Air India": 150
    }
    
    agency_cost = agency_surcharge.get(row["Flight_agency"], 100)
    
    # Departure time effect (peak hours: 6 AM - 10 AM & 6 PM - 10 PM increase price)
    departure_hour = int(row["Departure Time"].split(":")[0])
    peak_hour_surcharge = 200 if (6 <= departure_hour < 10 or 18 <= departure_hour < 22) else 0

    # Final price calculation
    price = base_price + (distance_cost + duration_cost) * multiplier + agency_cost + peak_hour_surcharge
    return round(price, 2)



df = India_datas.drop(columns={'Flight_price'})
# Apply function to calculate price
df["Calculated_Flight_Price"] = df.apply(calculate_flight_price, axis=1)

# Save updated file
df.to_excel("Updated_Flight_Data.xlsx", index=False)

print("Flight price calculation completed and saved!")
```

```python
import pandas as pd

def filter_round_trips(file_path, sheet_name="Sheet1"):
    # Load the Excel file
    df = pd.read_excel(file_path, sheet_name=sheet_name)
    
    # Creating a set of tuples for user travel history (Departure -> Arrival)
    user_trips = set(zip(df["User_ID"], df["Departure"], df["Arrival"]))
    
    # Identifying round trips (Arrival -> Departure exists for the same user)
    round_trips = {(user, arr, dep) for (user, dep, arr) in user_trips if (user, arr, dep) in user_trips}
    
    # Filtering out round trips from the dataframe
    filtered_df = df[~df.apply(lambda row: (row["User_ID"], row["Departure"], row["Arrival"]) in round_trips, axis=1)]
    
    return filtered_df

# Example usage
file_path = "df_INDIA.xlsx"  # Update this with the actual file path
filtered_data = filter_round_trips(file_path)

# Save the filtered data
filtered_data.to_excel("Filter_Flight_Data.xlsx", index=False)

print("Filtered data has been saved to 'Filter_Flight_Data.xlsx'")
```

```python
def calculate_flight_price(row):
    base_price = 500  # Base price for any flight

    # Adjusting price based on flight type
    type_multiplier = {
        "economic": 1.0,
        "premium": 1.3,
        "business": 1.6,
        "firstClass": 2.0
    }
    
    flight_type = row["flightType"]
    multiplier = type_multiplier.get(flight_type, 1.0)
    
    # Distance-based cost (per km charge)
    distance_cost = row["Flight Distance (km)"] * 2
    
    # Duration-based cost (per minute charge)
    duration_cost = convert_duration_to_minutes(row["Flight Duration (hh:mm)"]) * 1.5
    
    # Airline surcharge (mock values for different airlines)
    agency_surcharge = {
        "IndiGo": 100,
        "SpiceJet": 120,
        "GoAir": 110,
        "Air India": 150
    }
    
    agency_cost = agency_surcharge.get(row["Flight_agency"], 100)
    
    # Departure time effect (peak hours: 6 AM - 10 AM & 6 PM - 10 PM increase price)
    departure_hour = int(row["Departure Time"].split(":")[0])
    peak_hour_surcharge = 200 if (6 <= departure_hour < 10 or 18 <= departure_hour < 22) else 0

    # Final price calculation
    price = base_price + (distance_cost + duration_cost) * multiplier + agency_cost + peak_hour_surcharge
    return round(price, 2)

# Load the dataset


# Apply function to calculate price
OneWayHetvi["Calculated_Flight_Price"] = OneWayHetvi.apply(calculate_flight_price, axis=1)

OneWayHetvi
```
