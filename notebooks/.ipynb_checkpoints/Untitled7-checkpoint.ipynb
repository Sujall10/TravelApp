```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('flights.csv')
df
```

```python
df = df.drop(columns={'travelCode','userCode','time','distance'})
```

```python
df
```

```python
start_date = pd.Timestamp("2025-04-01")
end_date = pd.Timestamp("2025-05-01")
```

```python
df["Departure_date"] = df["date"].apply(lambda _: start_date + pd.to_timedelta(np.random.randint(0, (end_date - start_date).days), unit='D'))
```

```python
df
```

```python
cities_name = ['Tirupati','Visakhapatnam','Vijayawada','Itanagar','Dhubri','Guwahati','Patna','Raipur','Bilaspur','Raigarh','Mopa','Dabolim','Ahmedabad','Surat','Vadodara','Hisar','Shimla','Kullu-Manali','Jamshedpur','Ranchi','Bengaluru','Mysuru','Thiruvananthapuram','Gwalior','Indore','Bhopal','Mumbai','Navi Mumbai','Nashik','Pune','Imphal','Shillong','Puri','Bhubaneswar','Ludhiana','Amritsar','Jaipur','Udaipur','Kota','Gangtok','Coimbatore','Chennai','Hyderabad','Khowai','Ayodhya','Varanasi','Prayagraj','Dehradun','Kolkata','Delhi NCR','Chandigarh','Puducherry','Diu','Srinagar']
```

```python
num_entries = len(df)
data = []
```

```python
import random 
for _ in range(num_entries):
    departure = random.choice(cities_name)
    arrival = random.choice([city for city in cities_name if city != departure])
    data.append((departure, arrival))
```

```python
df_new = pd.DataFrame(data, columns=["Departure", "Arrival"])
df_new
```

```python
df = pd.concat([df, df_new], axis=1)
```

```python
df
```

```python
df_loc = pd.read_excel("AirportCoordinates.xlsx")
df_loc
```

```python
code_to_name = dict(zip(df_loc['City'], df_loc['Name']))
df['Departure'] = df['Departure'].map(code_to_name)
df['Arrival'] = df['Arrival'].map(code_to_name)
```

```python
airport_df = pd.read_excel("AirportCoordinates.xlsx")
```

```python
df = df.drop(columns={'from','to','price','date'})
```

```python
a = [
"Akasa Air",
"IndiGo",      
"Air India"   ,     
"SpiceJet"     ,    
"AirAsia India" ,   
"GoAir"
]
```

```python
df['agency'].value_counts()
```

```python
weights = [3, 3, 3, 3, 1, 1]
```

```python
df["agency"] = random.choices(a, weights=weights, k=len(df))
```

```python
df['agency'].value_counts()
```

```python
def random_time():
    hour = random.randint(0, 23)  # Random hour (0-23)
    minute = random.randint(0, 59)  # Random minute (0-59)
    return f"{hour:02d}:{minute:02d}"
```

```python
df["Departure_time"] = [random_time() for _ in range(len(df))]
```

```python
df.info()
```

```python
df['Departure_date'] = df['Departure_date'] + pd.to_timedelta(df['Departure_time'] + ":00")
```
